# Memory-decay FDR score

The memory-decay FDR score emphasizes recent discoveries more than older discoveries.

This cell scores a fixed sequence of decisions and ground-truth labels to show how the metric changes over time.

```python
from online_fdr.core.utils.evaluation import MemoryDecayFDR

rejects = [False, True, False, True, True, False, True, False]
alternatives = [False, True, False, False, True, False, True, False]
score = MemoryDecayFDR(delta=0.90)

rows = []
for idx, (reject, is_alternative) in enumerate(zip(rejects, alternatives), start=1):
    current_score = score.score_one(reject, is_alternative)
    rows.append((idx, reject, is_alternative, current_score))

print("Memory-decay FDR score")
print("delta=0.90")
print("")
print("step  reject  alternative  score")
for idx, reject, is_alternative, current_score in rows:
    print(
        f"{idx:>4}  {str(reject):>6}  {str(is_alternative):>11}  {current_score:>5.3f}"
    )
print("")
print(f"Final memory-decay FDR score: {rows[-1][3]:.3f}")
```
