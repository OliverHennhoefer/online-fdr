# ADDIS

ADDIS is an online p-value method that can discard large p-values before updating its adaptive threshold.

This cell runs ADDIS on a short p-value stream and reports the threshold used at each step.

```python
from online_fdr.p_values import Addis

p_values = [0.20, 0.004, 0.32, 0.018, 0.60, 0.0007, 0.041, 0.22]
method = Addis(alpha=0.05, wealth=0.025, lambda_=0.25, tau=0.50)

rows = []
for idx, p_value in enumerate(p_values, start=1):
    reject = method.test_one(p_value)
    rows.append((idx, p_value, method.last_rejection_threshold, reject))

print("ADDIS sequential p-value test")
print("target_level=0.05, lambda=0.25, tau=0.50")
print("")
print("test  p-value  threshold  reject")
for idx, p_value, threshold, reject in rows:
    print(f"{idx:>4}  {p_value:>7.4f}  {threshold:>9.6f}  {str(reject):>6}")
print("")
print(f"Tests processed: {method.num_hypotheses}")
print(f"Discoveries: {sum(row[3] for row in rows)}")
```
