# LORD with memory decay

LORD with memory decay down-weights older discoveries when setting new thresholds.

This cell runs the method on a short p-value stream and tracks the memory-decay FDR score for this simulated sequence.

```python
from online_fdr.core.utils import MemoryDecayFDR
from online_fdr.p_values import LORDMemoryDecay

p_values = [0.003, 0.12, 0.004, 0.02, 0.0008, 0.40, 0.006, 0.03]
labels = [True, False, True, False, True, False, True, False]
method = LORDMemoryDecay(alpha=0.20, delta=0.95, eta=1.0)
score = MemoryDecayFDR(delta=0.95)

rows = []
for idx, (p_value, is_alternative) in enumerate(zip(p_values, labels), start=1):
    reject = method.test_one(p_value)
    current_score = score.score_one(reject, is_alternative)
    rows.append((idx, p_value, method.last_rejection_threshold, reject, current_score))

print("LORD memory-decay p-value test")
print("target_level=0.20, delta=0.95, eta=1.00")
print("")
print("test  p-value  threshold  reject  memory_fdr")
for idx, p_value, threshold, reject, current_score in rows:
    print(
        f"{idx:>4}  {p_value:>7.4f}  {threshold:>9.6f}  {str(reject):>6}  {current_score:>10.3f}"
    )
print("")
print(f"Tests processed: {method.num_hypotheses}")
print(f"Discoveries: {sum(row[3] for row in rows)}")
print(f"Final memory-decay FDR score: {rows[-1][4]:.3f}")
```
