# Alpha spending

Alpha spending allocates a fixed error budget across a planned sequence of tests.

This cell applies Bonferroni spending, so each p-value is compared with the same planned threshold.

```python
from online_fdr.p_values import AlphaSpending
from online_fdr.p_values.spending.functions.bonferroni import Bonferroni

p_values = [0.030, 0.004, 0.120, 0.009, 0.0007, 0.080, 0.011, 0.400]
method = AlphaSpending(alpha=0.05, spend_func=Bonferroni(k=len(p_values)))

rows = []
for idx, p_value in enumerate(p_values, start=1):
    reject = method.test_one(p_value)
    rows.append((idx, p_value, method.current_threshold, reject))

print("Alpha spending p-value test")
print("rule=Bonferroni, target_alpha=0.05")
print("")
print("test  p-value  threshold  reject")
for idx, p_value, threshold, reject in rows:
    print(f"{idx:>4}  {p_value:>7.4f}  {threshold:>9.6f}  {str(reject):>6}")
print("")
print(f"Tests processed: {method.num_tests}")
print(f"Discoveries: {sum(row[3] for row in rows)}")
```
