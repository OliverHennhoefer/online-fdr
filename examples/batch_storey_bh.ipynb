# Batch Storey-BH

Batch Storey-BH tests a group of p-values at the same online step.

This cell submits one batch and prints the batch threshold and decisions in input order.

```python
from online_fdr.p_values import BatchStoreyBH

p_values = [0.001, 0.012, 0.20, 0.40, 0.006, 0.70]
method = BatchStoreyBH(alpha=0.10, lambda_=0.50)

decisions = method.test_batch(p_values)

print("Batch Storey-BH p-value test")
print(f"target_fdr=0.10, lambda=0.50, batch_threshold={method.current_threshold:.6f}")
print("")
print("index  p-value  reject")
for idx, (p_value, reject) in enumerate(zip(p_values, decisions), start=1):
    print(f"{idx:>5}  {p_value:>7.4f}  {str(reject):>6}")
print("")
print(f"Batches processed: {method.num_tests}")
print(f"Discoveries in batch: {sum(decisions)} of {len(decisions)}")
```
